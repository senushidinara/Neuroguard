## Process/analyze data

### Subtask:
Process and analyze the loaded EDF data.

**Reasoning**:
Iterate through the loaded EDF data, extract signal information and data, store it, and perform a preliminary analysis by examining the number of signals, types, sampling rates, and visualizing a small portion.

```python
import numpy as np
import matplotlib.pyplot as plt
import os

extracted_edf_data = []

print(f"\nProcessing data from {len(loaded_edf_data)} loaded EDF files:")

for file_path, edf_reader in loaded_edf_data.items():
    try:
        num_signals = edf_reader.signals_in_file
        signal_headers = edf_reader.getSignalHeaders()
        file_data = {
            'file_path': file_path,
            'num_signals': num_signals,
            'signal_headers': signal_headers,
            'signals_data': {}
        }

        print(f"\nProcessing file: {file_path}")
        print(f"Number of signals: {num_signals}")

        # Extract data for each signal
        for i in range(num_signals):
            signal_header = signal_headers[i]
            signal_label = signal_header['label']
            # Corrected access to sampling_frequency
            sampling_frequency = signal_header.get('sampling_frequency', 0)
            physical_dimension = signal_header.get('physical_dimension', '')

            print(f"  Signal {i+1}: Label='{signal_label}', Fs={sampling_frequency} Hz, Units='{physical_dimension}'")

            # Read a portion of the signal data for preliminary analysis/visualization
            # Reading the first 10 seconds of data (assuming Fs > 0)
            if sampling_frequency > 0:
                num_samples_to_read = int(sampling_frequency * 10)
                # Ensure we don't read more samples than available
                total_samples = edf_reader.getNSamples()[i]
                num_samples_to_read = min(num_samples_to_read, total_samples)

                if num_samples_to_read > 0:
                    signal_data = edf_reader.readsignal(i, 0, num_samples_to_read)
                    file_data['signals_data'][signal_label] = signal_data
                else:
                     print(f"  No samples to read for signal '{signal_label}'.")
                     file_data['signals_data'][signal_label] = np.array([]) # Store empty array
            else:
                print(f"  Skipping data read for signal '{signal_label}' due to zero or invalid sampling frequency.")
                file_data['signals_data'][signal_label] = np.array([]) # Store empty array


        # Only append file_data if it contains actual signal data
        if file_data['signals_data']:
             extracted_edf_data.append(file_data)
        else:
             print(f"  No signal data extracted for {file_path}. Skipping.")


    except Exception as e:
        print(f"Error processing data from {file_path}: {e}")

print("\nPreliminary Analysis Summary:")
print(f"Processed data from {len(extracted_edf_data)} files with extracted signal data.")

# Example visualization of a few signals from the first processed file (if available)
if extracted_edf_data:
    first_file_data = extracted_edf_data[0]
    file_path = first_file_data['file_path']
    signals_data = first_file_data['signals_data']
    signal_headers = first_file_data['signal_headers']


    print(f"\nVisualizing a portion of data from: {file_path}")

    plt.figure(figsize=(15, 8))
    offset = 0
    plotted_signals = 0 # Keep track of how many signals are plotted

    for signal_label, data in signals_data.items():
        if data.size > 0:
            # Find the corresponding signal header to get sampling frequency
            sampling_frequency = 0
            for header in signal_headers:
                if header['label'] == signal_label:
                    sampling_frequency = header.get('sampling_frequency', 0)
                    break

            if sampling_frequency > 0:
                time = np.arange(len(data)) / sampling_frequency
                plt.plot(time, data + offset, label=signal_label)
                offset += np.max(np.abs(data)) * 2 if data.size > 0 else 0 # Add offset to separate signals
                plotted_signals += 1
            else:
                print(f"  Cannot plot signal '{signal_label}' due to zero sampling frequency.")
        else:
            print(f"  No data to visualize for signal: {signal_label}")

    if plotted_signals > 0:
        plt.title(f"Sample Signals from {os.path.basename(file_path)}")
        plt.xlabel("Time (s)")
        plt.ylabel("Amplitude + Offset")
        plt.legend()
        plt.grid(True)
        plt.show()
    else:
        print(f"  No signals with data were plotted for {os.path.basename(file_path)}.")

else:
    print("\nNo data was successfully extracted to perform visualization.")
```

```python
import pingouin as pg
import pandas as pd
import numpy as np

print("\nCalculating ICC[2,1] for Sleep Statistics:")
icc_sleep_stats = {}

# Identify metrics with data from multiple nights in sleep_stats_icc_df
# We need at least two non-null values across the 'nights' for each subject for a metric
# Assuming columns like 'E0_Metric', 'EC_Metric', etc.
# We can group columns by the base metric name (e.g., TST_sec, REM_duration_sec)

# Get base metric names (remove the night prefix like E0_, EC_)
sleep_stats_metrics = set([col.split('_', 1)[1] for col in sleep_stats_icc_df.columns if '_' in col])

for metric in sleep_stats_metrics:
    # Select columns for this metric across all nights
    metric_columns = [col for col in sleep_stats_icc_df.columns if col.endswith(f'_{metric}')]

    if len(metric_columns) >= 2: # Need at least two nights for ICC
        # Create a temporary DataFrame for this metric with subjects as rows and nights as columns
        temp_df = sleep_stats_icc_df[metric_columns]

        # Rename columns to just the night info (e.g., E0, EC)
        temp_df.columns = [col.split('_')[0] for col in metric_columns]

        # Ensure there are at least two non-null values for each subject across nights for valid ICC
        # Pingouin's icc function handles NaNs, but it's good to be aware of the data structure.
        # For ICC(2,1), we need subjects (rows) and 'raters'/'measurements' (columns - nights).
        # The input to pg.icc is typically a DataFrame where each column is a 'rater' or 'measurement'.
        # Our temp_df is already in this format (subjects x nights).

        # Perform ICC calculation
        try:
            # The 'data' parameter for pg.icc is the DataFrame.
            # 'subjects' parameter is the column/index representing subjects (which is our index).
            # 'ratings' parameter represents the measurements (our columns, which are nights).
            # ICC(2,1) is Type 2, single rater/measurement.

            icc_result = pg.icc(data=temp_df, targets='Subject', raters='columns', subject='Subject', inplace=False).round(3)

            # Extract the ICC[2,1] value (Type = 'ICC2', subtype = 'Single')
            icc_value = icc_result[(icc_result['Type'] == 'ICC2') & (icc_result['Estimator'] == 'Single')]['ICC'].values[0]
            p_value = icc_result[(icc_result['Type'] == 'ICC2') & (icc_result['Estimator'] == 'Single')]['p-val'].values[0]
            ci_lower = icc_result[(icc_result['Type'] == 'ICC2') & (icc_result['Estimator'] == 'Single')]['CI95%'].values[0][0]
            ci_upper = icc_result[(icc_result['Type'] == 'ICC2') & (icc_result['Estimator'] == 'Single')]['CI95%'].values[0][1]


            icc_sleep_stats[metric] = {
                'ICC': icc_value,
                'p-val': p_value,
                'CI95%': (ci_lower, ci_upper)
            }
            print(f"  {metric}: ICC = {icc_value:.3f}, p = {p_value:.3f}, 95% CI = [{ci_lower:.3f}, {ci_upper:.3f}]")


        except Exception as e:
            print(f"  Error calculating ICC for {metric}: {e}")


print("\nFinished calculating ICC for Sleep Statistics.")

print("\nCalculating ICC[2,1] for Spectral Features:")
icc_spectral_features = {}

# Identify metrics with data from multiple nights in spectral_features_icc_df
# Get base metric names (remove the night prefix like E0_)
spectral_features_metrics = set([col.split('_', 1)[1] for col in spectral_features_icc_df.columns if '_' in col])

for metric in spectral_features_metrics:
    # Select columns for this metric across all nights
    metric_columns = [col for col in spectral_features_icc_df.columns if col.endswith(f'_{metric}')]

    if len(metric_columns) >= 2: # Need at least two nights for ICC
         # Create a temporary DataFrame for this metric with subjects as rows and nights as columns
        temp_df = spectral_features_icc_df[metric_columns]

        # Rename columns to just the night info (e.g., E0, EC)
        temp_df.columns = [col.split('_')[0] for col in metric_columns]

        # Perform ICC calculation
        try:
            icc_result = pg.icc(data=temp_df, targets='Subject', raters='columns', subject='Subject', inplace=False).round(3)

            # Extract the ICC[2,1] value (Type = 'ICC2', subtype = 'Single')
            icc_value = icc_result[(icc_result['Type'] == 'ICC2') & (icc_result['Estimator'] == 'Single')]['ICC'].values[0]
            p_value = icc_result[(icc_result['Type'] == 'ICC2') & (icc_result['Estimator'] == 'Single')]['p-val'].values[0]
            ci_lower = icc_result[(icc_result['Type'] == 'ICC2') & (icc_result['Estimator'] == 'Single')]['CI95%'].values[0][0]
            ci_upper = icc_result[(icc_result['Type'] == 'ICC2') & (icc_result['Estimator'] == 'Single')]['CI95%'].values[0][1]


            icc_spectral_features[metric] = {
                'ICC': icc_value,
                'p-val': p_value,
                'CI95%': (ci_lower, ci_upper)
            }
            print(f"  {metric}: ICC = {icc_value:.3f}, p = {p_value:.3f}, 95% CI = [{ci_lower:.3f}, {ci_upper:.3f}]")


        except Exception as e:
            print(f"  Error calculating ICC for {metric}: {e}")


print("\nFinished calculating ICC for Spectral Features.")
```

Organize the data so that each row represents a subject and each column represents a measurement from a different night.

**Reasoning**:
Process the `sleep_stats_list` and `spectral_features_list` to create pandas DataFrames where rows are subjects and columns are metrics from different nights, preparing the data for ICC calculation.

```python
import pandas as pd
import os

# Assuming sleep_stats_list and spectral_features_list contain the calculated metrics

# Structure Sleep Statistics Data
print("\nStructuring Sleep Statistics Data for ICC:")
sleep_stats_icc_data = {}

for stats in sleep_stats_list:
    filename = stats['filename']
    # Extract subject ID and night information from the filename (assuming SCxxxxE0-PSG.edf format)
    # Example: SC4001E0-PSG.edf -> Subject: SC4001, Night: E0
    parts = os.path.basename(filename).split('-')
    if len(parts) >= 2:
        subject_id = parts[0][:5] # Assuming SCxxxx format for subject ID
        night_info = parts[0][5:] # Assuming E0, EC, EH, etc. is night info
        metric_prefix = f"{night_info}_" # Prefix for column names (e.g., E0_REM_duration)

        if subject_id not in sleep_stats_icc_data:
            sleep_stats_icc_data[subject_id] = {}

        # Add sleep statistics metrics to the dictionary
        sleep_stats_icc_data[subject_id][f"{metric_prefix}TST_sec"] = stats['total_sleep_time_sec']
        sleep_stats_icc_data[subject_id][f"{metric_prefix}REM_duration_sec"] = stats['rem_duration_sec']
        sleep_stats_icc_data[subject_id][f"{metric_prefix}REM_proportion_tst"] = stats['rem_proportion_tst']

        # Add other stage durations (optional, but useful for comprehensive ICC)
        for stage, duration in stats['stage_durations_sec'].items():
             # Sanitize stage names for column names (replace spaces and special characters)
             sanitized_stage = stage.replace(' ', '_').replace('-', '_').replace('?', 'Unknown')
             sleep_stats_icc_data[subject_id][f"{metric_prefix}{sanitized_stage}_duration_sec"] = duration


# Convert the dictionary to a pandas DataFrame
sleep_stats_icc_df = pd.DataFrame.from_dict(sleep_stats_icc_data, orient='index')
sleep_stats_icc_df.index.name = 'Subject'

print("\nSleep Statistics DataFrame for ICC:")
display(sleep_stats_icc_df)


# Structure Spectral Features Data
print("\nStructuring Spectral Features Data for ICC:")
spectral_features_icc_data = {}

for features in spectral_features_list:
    filename = features['filename']
    # Extract subject ID and night information from the filename (assuming SCxxxxE0-PSG.edf format)
    parts = os.path.basename(filename).split('-')
    if len(parts) >= 2:
        subject_id = parts[0][:5] # Assuming SCxxxx format for subject ID
        night_info = parts[0][5:] # Assuming E0, EC, EH, etc. is night info
        metric_prefix = f"{night_info}_" # Prefix for column names (e.g., E0_frontal_theta_gamma_ratio)

        if subject_id not in spectral_features_icc_data:
            spectral_features_icc_data[subject_id] = {}

        # Add spectral features metrics to the dictionary
        spectral_features_icc_data[subject_id][f"{metric_prefix}frontal_theta_power"] = features['frontal_theta_power']
        spectral_features_icc_data[subject_id][f"{metric_prefix}frontal_gamma_power"] = features['frontal_gamma_power']
        spectral_features_icc_data[subject_id][f"{metric_prefix}frontal_theta_gamma_ratio"] = features['frontal_theta_gamma_ratio']
        spectral_features_icc_data[subject_id][f"{metric_prefix}occipital_theta_power"] = features['occipital_theta_power']
        spectral_features_icc_data[subject_id][f"{metric_prefix}occipital_gamma_power"] = features['occipital_gamma_power']
        spectral_features_icc_data[subject_id][f"{metric_prefix}occipital_theta_gamma_ratio"] = features['occipital_theta_gamma_ratio']


# Convert the dictionary to a pandas DataFrame
spectral_features_icc_df = pd.DataFrame.from_dict(spectral_features_icc_data, orient='index')
spectral_features_icc_df.index.name = 'Subject'

print("\nSpectral Features DataFrame for ICC:")
display(spectral_features_icc_df)

print("\nFinished structuring data for ICC calculation.")
```

## Calculate sleep statistics

### Subtask:
Use YASA to calculate sleep statistics, including REM duration and proportion of total sleep time.

**Reasoning**:
Iterate through the extracted annotations data, count the occurrences and durations of each sleep stage, calculate total sleep time and percentages, and print the summary.

```python
from collections import Counter
import numpy as np

sleep_stage_counts = {}
sleep_stage_durations = {}
total_sleep_time_all_subjects = 0

print("\nAnalyzing sleep stages and durations:")

for filename, annotations in extracted_annotations_data.items():
    print(f"\nAnalyzing annotations for: {os.path.basename(filename)}")

    # Count occurrences and calculate duration for each sleep stage in the current file
    current_stage_counts = Counter()
    current_stage_durations = Counter() # Store durations in seconds

    for annot in annotations:
        description = annot['description']
        duration = annot['duration']

        # Accumulate counts and durations
        current_stage_counts[description] += 1
        current_stage_durations[description] += duration

    # Print summary for the current file
    print("  Sleep Stage Counts:")
    for stage, count in current_stage_counts.items():
        print(f"    {stage}: {count} occurrences")

    print("  Sleep Stage Durations (seconds):")
    for stage, duration in current_stage_durations.items():
        print(f"    {stage}: {duration:.2f} s")

    # Aggregate counts and durations across all files
    for stage, count in current_stage_counts.items():
        sleep_stage_counts[stage] = sleep_stage_counts.get(stage, 0) + count

    for stage, duration in current_stage_durations.items():
        sleep_stage_durations[stage] = sleep_stage_durations.get(stage, 0) + duration

# Calculate total sleep time across all subjects
# Sleep stages typically include Wake, REM, N1, N2, N3 (or S1, S2, S3, S4)
# Let's assume "Wake" and "Movement time" are not considered "sleep time" for this calculation.
# You might need to adjust this based on the specific descriptions in your annotations.
sleep_stages_for_total = ['Sleep stage 1', 'Sleep stage 2', 'Sleep stage 3', 'Sleep stage 4', 'REM sleep']
# Also include 'N1', 'N2', 'N3' if those descriptions are present
sleep_stages_for_total.extend(['N1', 'N2', 'N3'])


total_sleep_time_all_subjects = sum(duration for stage, duration in sleep_stage_durations.items() if stage in sleep_stages_for_total)

print(f"\nTotal sleep time across all subjects: {total_sleep_time_all_subjects:.2f} seconds")
print(f"Total sleep time across all subjects: {total_sleep_time_all_subjects / 3600:.2f} hours")


# Calculate percentage of total sleep time for each sleep stage
sleep_stage_percentages = {}
if total_sleep_time_all_subjects > 0:
    for stage, duration in sleep_stage_durations.items():
        if stage in sleep_stages_for_total:
            percentage = (duration / total_sleep_time_all_subjects) * 100
            sleep_stage_percentages[stage] = percentage
else:
    print("\nTotal sleep time is zero, cannot calculate percentages.")

# Print combined summary across all subjects
print("\nCombined Sleep Stage Summary (All Subjects):")
print("  Total Occurrences:")
for stage, count in sleep_stage_counts.items():
    print(f"    {stage}: {count}")

print("  Total Durations (seconds):")
for stage, duration in sleep_stage_durations.items():
     print(f"    {stage}: {duration:.2f} s ({duration / 3600:.2f} hours)")


if sleep_stage_percentages:
    print("  Percentage of Total Sleep Time:")
    # Sort by percentage for better readability
    sorted_percentages = sorted(sleep_stage_percentages.items(), key=lambda item: item[1], reverse=True)
    for stage, percentage in sorted_percentages:
        print(f"    {stage}: {percentage:.2f}%")

print("\nAnalysis complete.")

# Optional: Explore other features from raw_data_list
print("\nExploring additional features from raw data objects:")
if raw_data_list:
    for i, raw in enumerate(raw_data_list):
        try:
            raw_filename = os.path.basename(raw._filenames[0]) if hasattr(raw, '_filenames') and raw._filenames else (raw.info['filename'] if hasattr(raw.info, 'filename') else f"Raw object {i+1}")
            print(f"\nInformation for Raw Data: {raw_filename}")
            print(f"  Duration: {raw.times[-1]:.2f} seconds ({raw.times[-1]/3600:.2f} hours)")
            print(f"  Number of channels: {len(raw.ch_names)}")
            print(f"  Channel types: {raw.get_channel_types()}")
            print(f"  Sampling frequency: {raw.info['sfreq']} Hz")
        except Exception as e:
            print(f"Error exploring raw data object {i+1}: {e}")
else:
    print("No raw data objects available for exploring additional features.")
```

## Calculate sleep statistics

### Subtask:
Use YASA to calculate sleep statistics, including REM duration and proportion of total sleep time.


**Reasoning**:
Iterate through the prepared data, calculate sleep statistics using YASA, and print the results.



```python
import yasa
import os

sleep_stats_list = []

print("\nCalculating sleep statistics using YASA:")

for raw in prepared_data_for_yasa:
    try:
        # Get the filename for reporting
        raw_filename = os.path.basename(raw._filenames[0]) if hasattr(raw, '_filenames') and raw._filenames else "Unknown file"
        print(f"\nAnalyzing sleep statistics for: {raw_filename}")

        # Calculate sleep statistics from the loaded annotations
        # YASA's SleepStaging class is typically used for predicting stages,
        # but its `get_wake_stages` and similar methods can be used to analyze existing annotations.
        # Alternatively, we can directly analyze the annotations object for durations.

        # Let's calculate statistics directly from the annotations for simplicity and to use existing results.
        annotations = raw.annotations

        # Get sleep stage durations
        stage_durations = annotations.to_data_frame().groupby('description')['duration'].sum().to_dict()

        # Calculate Total Sleep Time (TST) - sum of durations for non-Wake and non-Movement stages
        # Including 'Sleep stage R' for REM sleep, and 'Sleep stage 1', 'Sleep stage 2', 'Sleep stage 3', 'Sleep stage 4' for NREM.
        sleep_stages_for_tst = ['Sleep stage 1', 'Sleep stage 2', 'Sleep stage 3', 'Sleep stage 4', 'Sleep stage R']
        # Also include 'N1', 'N2', 'N3', 'R' if those descriptions are present
        sleep_stages_for_tst.extend(['N1', 'N2', 'N3']) # Assuming 'N1', 'N2', 'N3' are NREM stages
        # Assuming 'R' is an alternative label for REM if 'REM sleep' is not used
        # Let's explicitly check for 'Sleep stage R' and 'REM sleep' for REM duration
        rem_stages = ['Sleep stage R', 'REM sleep']

        tst = sum(stage_durations.get(stage, 0) for stage in sleep_stages_for_tst)

        # Get REM duration
        rem_duration = sum(stage_durations.get(stage, 0) for stage in rem_stages)


        # Calculate REM proportion of TST
        rem_proportion_tst = (rem_duration / tst) * 100 if tst > 0 else 0

        # Store the statistics
        sleep_stats = {
            'filename': raw_filename,
            'total_sleep_time_sec': tst,
            'total_sleep_time_hours': tst / 3600,
            'rem_duration_sec': rem_duration,
            'rem_duration_hours': rem_duration / 3600,
            'rem_proportion_tst': rem_proportion_tst,
            'stage_durations_sec': stage_durations # Store all stage durations
        }
        sleep_stats_list.append(sleep_stats)

        # Print summary for the current file
        print(f"  Total Sleep Time: {sleep_stats['total_sleep_time_hours']:.2f} hours ({sleep_stats['total_sleep_time_sec']:.2f} seconds)")
        print(f"  REM Sleep Duration: {sleep_stats['rem_duration_hours']:.2f} hours ({sleep_stats['rem_duration_sec']:.2f} seconds)")
        print(f"  REM Proportion of TST: {sleep_stats['rem_proportion_tst']:.2f}%")
        print("  All Stage Durations (seconds):")
        for stage, duration in sleep_stats['stage_durations_sec'].items():
            print(f"    {stage}: {duration:.2f} s")


    except Exception as e:
        print(f"Error calculating sleep statistics for {raw_filename}: {e}")

print("\nFinished calculating sleep statistics.")

# Print combined summary across all subjects (optional, but good for overview)
if sleep_stats_list:
    print("\nCombined Sleep Statistics Summary (All Subjects):")
    total_tst_all = sum(stats['total_sleep_time_sec'] for stats in sleep_stats_list)
    total_rem_all = sum(stats['rem_duration_sec'] for stats in sleep_stats_list)
    total_rem_proportion_all = (total_rem_all / total_tst_all) * 100 if total_tst_all > 0 else 0

    print(f"  Average Total Sleep Time: {np.mean([stats['total_sleep_time_hours'] for stats in sleep_stats_list]):.2f} hours")
    print(f"  Total Total Sleep Time (All Subjects): {total_tst_all / 3600:.2f} hours")
    print(f"  Total REM Sleep Duration (All Subjects): {total_rem_all / 3600:.2f} hours")
    print(f"  Overall REM Proportion of TST (All Subjects): {total_rem_proportion_all:.2f}%")

    # Calculate average duration for each stage across subjects
    avg_stage_durations = Counter()
    for stats in sleep_stats_list:
        for stage, duration in stats['stage_durations_sec'].items():
             avg_stage_durations[stage] += duration

    print("  Average Stage Durations (seconds):")
    num_subjects_with_stage = Counter()
    for stats in sleep_stats_list:
         for stage in stats['stage_durations_sec'].keys():
              num_subjects_with_stage[stage] += 1

    for stage, total_duration in avg_stage_durations.items():
        if num_subjects_with_stage[stage] > 0:
             print(f"    {stage}: {total_duration / num_subjects_with_stage[stage]:.2f} s")
        else:
             print(f"    {stage}: 0.00 s (No subjects with this stage)")


else:
    print("No sleep statistics were calculated.")
```

```python
import yasa
import numpy as np
import matplotlib.pyplot as plt
import os
import mne

spectral_features_list = []

print("\nCalculating spectral features (Theta/Gamma Ratios) using MNE and YASA:")

for raw in prepared_data_for_yasa:
    try:
        # Get the filename for reporting
        raw_filename = os.path.basename(raw._filenames[0]) if hasattr(raw, '_filenames') and raw._filenames else "Unknown file"
        print(f"\nAnalyzing spectral features for: {raw_filename}")

        # Select channels for frontal and occipital regions
        # Based on the previous output, channel labels are like 'EEG Fpz-Cz' and 'EEG Pz-Oz'
        frontal_channels = mne.pick_channels(raw.ch_names, include=['EEG Fpz-Cz'], exclude=[])
        occipital_channels = mne.pick_channels(raw.ch_names, include=['EEG Pz-Oz'], exclude=[])

        if not frontal_channels:
            print(f"  Warning: No frontal channels found for {raw_filename}. Skipping spectral analysis for frontal.")
        if not occipital_channels:
            print(f"  Warning: No occipital channels found for {raw_filename}. Skipping spectral analysis for occipital.")

        file_spectral_features = {
            'filename': raw_filename,
            'frontal_theta_power': None,
            'frontal_gamma_power': None,
            'frontal_theta_gamma_ratio': None,
            'occipital_theta_power': None,
            'occipital_gamma_power': None,
            'occipital_theta_gamma_ratio': None,
        }


        # Calculate PSD for frontal channels
        if frontal_channels:
            try:
                # Calculate PSD using multitaper method
                psds_frontal, freqs_frontal = mne.time_frequency.psd_multitaper(raw, picks=frontal_channels, fmin=0.5, fmax=40., verbose=False)
                psds_frontal = psds_frontal.mean(0) # Average across channels if multiple frontal were picked

                # Define frequency bands
                theta_band = (4, 8) # Hz
                gamma_band = (30, 40) # Hz

                # Calculate power in bands
                theta_power_frontal = yasa.bandpower(psds_frontal, freqs_frontal, band=theta_band).values[0]
                gamma_power_frontal = yasa.bandpower(psds_frontal, freqs_frontal, band=gamma_band).values[0]

                # Calculate theta/gamma ratio
                theta_gamma_ratio_frontal = theta_power_frontal / gamma_power_frontal if gamma_power_frontal > 0 else np.nan

                file_spectral_features['frontal_theta_power'] = theta_power_frontal
                file_spectral_features['frontal_gamma_power'] = gamma_power_frontal
                file_spectral_features['frontal_theta_gamma_ratio'] = theta_gamma_ratio_frontal

                print(f"  Frontal (EEG Fpz-Cz): Theta Power={theta_power_frontal:.4f}, Gamma Power={gamma_power_frontal:.4f}, Theta/Gamma Ratio={theta_gamma_ratio_frontal:.4f}")

            except Exception as e:
                print(f"  Error calculating spectral features for frontal channels in {raw_filename}: {e}")


        # Calculate PSD for occipital channels
        if occipital_channels:
            try:
                # Calculate PSD using multitaper method
                psds_occipital, freqs_occipital = mne.time_frequency.psd_multitaper(raw, picks=occipital_channels, fmin=0.5, fmax=40., verbose=False)
                psds_occipital = psds_occipital.mean(0) # Average across channels if multiple occipital were picked

                # Define frequency bands (using the same bands as frontal)
                theta_band = (4, 8) # Hz
                gamma_band = (30, 40) # Hz

                # Calculate power in bands
                theta_power_occipital = yasa.bandpower(psds_occipital, freqs_occipital, band=theta_band).values[0]
                gamma_power_occipital = yasa.bandpower(psds_occipital, freqs_occipital, band=gamma_band).values[0]

                # Calculate theta/gamma ratio
                theta_gamma_ratio_occipital = theta_power_occipital / gamma_power_occipital if gamma_power_occipital > 0 else np.nan

                file_spectral_features['occipital_theta_power'] = theta_power_occipital
                file_spectral_features['occipital_gamma_power'] = gamma_power_occipital
                file_spectral_features['occipital_theta_gamma_ratio'] = theta_gamma_ratio_occipital

                print(f"  Occipital (EEG Pz-Oz): Theta Power={theta_power_occipital:.4f}, Gamma Power={gamma_power_occipital:.4f}, Theta/Gamma Ratio={theta_gamma_ratio_occipital:.4f}")

            except Exception as e:
                print(f"  Error calculating spectral features for occipital channels in {raw_filename}: {e}")

        spectral_features_list.append(file_spectral_features)


    except Exception as e:
        print(f"Error processing {raw_filename} for spectral analysis: {e}")

print("\nFinished calculating spectral features.")

# Print summary of spectral features
print("\nSpectral Features Summary:")
for features in spectral_features_list:
    print(f"\nFile: {features['filename']}")
    if features['frontal_theta_gamma_ratio'] is not None:
        print(f"  Frontal Theta/Gamma Ratio: {features['frontal_theta_gamma_ratio']:.4f}")
    else:
        print("  Frontal Theta/Gamma Ratio: N/A (Channels not found or error)")

    if features['occipital_theta_gamma_ratio'] is not None:
        print(f"  Occipital Theta/Gamma Ratio: {features['occipital_theta_gamma_ratio']:.4f}")
    else:
        print("  Occipital Theta/Gamma Ratio: N/A (Channels not found or error)")

print("\nSpectral analysis complete.")
```
